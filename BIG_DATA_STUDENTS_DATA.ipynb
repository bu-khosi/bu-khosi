```python
#how to correct the data types#

df = spark.read.option('header',True).option('inferSchema',True).csv(Data_path)

df.show()
```

```python
#changing datatypes....for practice purposes#
from pyspark.sql.functions import col

df =df.withColumn('roll', col('roll').cast('string'))

df.printSchema()
```

```python
df.printSchema()
#shows the corrected data types for each attribute#
```

```python
from pyspark.sql.functions import col,lit
```

```python
df = df.withColumn('roll', col('roll').cast('string'))

df.printSchema()
```

```python
df.show(n=5)
```

```python
#modify column value#

df = df.withColumn('marks',col('marks') +10)
df.show()
```

```python

df = df.withColumn('marks',col('marks')/10 )
df.show()
```

```python
#when you want to add a new column#
df =df.withColumn('country_name',lit('south_africa'))
df.show()
```

```python
#renaming a column#
df = df.withColumnRenamed('course opted','course')
df.show()
```

```python
#renaming columns#

df = df.withColumnRenamed('gender','sex').withColumnRenamed('roll','roll_number')
df.show()
```

```python
#accessing 1 column#
df.select(col('name')).show()
```

```python
#FETCHING RECORDS BASED ON CONDITION USING filter()
# FETCH ALL RECORDS WHO HAVE OPTED DB AS A COURSE

df.filter(df['marks'] >=60).show()
```
